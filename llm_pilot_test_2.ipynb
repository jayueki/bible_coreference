### Prompt:
From sample_test5, classify the relationship between the **'From Verse'** and the **'To Verse'** for every row.
The goal is primarily to identify **strong, textually meaningful biblical cross-references**, especially places where one passage directly quotes, intentionally paraphrases, or directly alludes to another. Mere similarity of topic or theme should **not** count as a direct reference.
#### Classification Scheme
Use the following three classes:
##### 1. Direct Reference
Assign **Class 1 — Direct Reference** when there is strong evidence that one passage intentionally draws on another passage.
This includes:
- **Direct quotations**: one passage explicitly quotes or substantially reproduces wording from the other passage.
- **Intentional paraphrases**: one passage intentionally restates or quotes the other in different wording. The correspondence should be specific enough to indicate intentional use of that passage, rather than merely expressing the same idea.
- **Direct allusions**: one passage intentionally invokes or fulfills the other without directly quoting it. This includes clear fulfillment of prophecy and other cases where distinctive wording, imagery, details, or context strongly identify the specific passage being invoked.
For **paraphrases and allusions**, intentionality is required. There must be strong evidence that the author, speaker, or—in the case of prophetic fulfillment—the text's presentation of divine intention deliberately connects the passages.
Do not require identical wording for Class 1. Genuine quotations may differ because of translation or textual tradition, and intentional paraphrases or direct allusions may use different wording. Class 1 should still be assigned when the evidence of intentional dependence on or invocation of the specific passage is strong.
A shared event, person, topic, doctrine, narrative context, or historical connection by itself is **not** sufficient for Class 1. Likewise, a passage referring back to an earlier event is not Class 1 unless it also intentionally invokes the earlier passage through quotation, paraphrase, or direct allusion.
##### 2. Indirect Reference
Assign **Class 2 — Indirect Reference** when there is a meaningful relationship between the passages, but the evidence is not strong or specific enough to establish a direct reference.
This includes:
- passages explaining or providing context for one another;
- thematic or conceptual parallels;
- similar teachings expressed;
- related events, people, institutions, laws, doctrines, symbols, or ideas;
- possible but uncertain paraphrases or allusions;
- broader parallels.
##### 3. Weak/No Reference
Assign **Class 3 — Weak/No Reference** when there is no substantial reason to treat the verses as a useful biblical cross-reference.
This includes:
- superficial similarities;
- very generic shared words;
- extremely broad thematic overlap;
- connections that require several speculative interpretive steps;
- accidental verbal resemblance;
- passages with no clear textual, literary, contextual, historical, or explanatory relationship.
Do not assign Class 2 merely because two passages can be connected under a sufficiently broad theological theme. The relationship should be useful and reasonably identifiable.
#### Important Decision Rule
Use the following hierarchy when deciding between classes:
**Quotation, intentional paraphrase, or direct intentional allusion → Class 1**
**Meaningful relationship without evidence of intentional invocation → Class 2**
**Superficial, speculative, or no meaningful relationship → Class 3**
For Class 1, ask: **Is there strong evidence that the author, speaker, or—in the case of prophetic fulfillment—the text's presentation of divine intention intentionally invokes this specific passage, rather than merely the same event, person, topic, or idea?**
If yes, assign Class 1. If the passages are clearly related but intentional invocation of the specific passage is uncertain, assign Class 2. Because the goal includes **high recall**, do not require exact verbal correspondence when there is otherwise strong evidence of intentional quotation, paraphrase, or direct allusion.
Ask more generally: **Is there evidence that these two specific passages belong together, rather than simply discussing similar things?**
If yes, determine whether the relationship is direct (Class 1) or indirect (Class 2). If the connection could be made equally well between many unrelated verses discussing the same broad subject, it is probably Class 3.
#### Output

- `Class`
- `Class Name`
- `Rationale`

- `1` = `Direct Reference`
- `2` = `Indirect Reference`

The **Rationale** should be concise but specific. State the actual evidence connecting the passages rather than simply restating the class definition.
#### Classification Standard
Be **conservative with Class 1 while maintaining high recall for genuine quotations, intentional paraphrases, and direct allusions**. The purpose of this task is to distinguish strong biblical cross-references from general semantic similarity.
Do not infer a direct reference merely because:
- the verses use a common biblical word;
- they concern the same broad doctrine;
- they refer to the same person or event;
- one passage explains or provides context for the other;
- they occur within the same narrative or historical sequence;
- commentators might discuss them together;

Reserve **Class 1** for direct quotations, intentional paraphrases, and direct intentional allusions where there is identifiable evidence that the specific passage is being invoked. For paraphrases and allusions, the evidence should support intentionality by the author, speaker, or—in the case of prophecy—the text's presentation of divine intention.
