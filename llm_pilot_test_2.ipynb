```python
a = test4b1_classified['Class'].astype(str)
b = test4b2_classified['Class'].astype(str)

diff = a != b

# Count individual classes involved in disagreements
controversial = Counter()

for x, y in zip(a[diff], b[diff]):
    classes = set(x.split(';')) ^ set(y.split(';'))
    controversial.update(classes)

controversial_df = (
    pd.DataFrame(controversial.items(), columns=['Class', 'Disagreements'])
      .sort_values('Disagreements', ascending=False)
)
```
