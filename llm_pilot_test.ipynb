# Test 2
Testing whether AI identify direct quotes: the previous random sample did not contain direct quotes, which we are most interested in. Thus, we include three direct quotes, to see if the model can accurately identify them out of the other edges.

```python
# known direct quotes:
# Matthew 4:4 -> Deuteronomy 8:3
# Matthew 21:42 -> Psalms 118:22-23
# Romans 1:17 -> Habakkuk 2:4
pairs = {
    'Matt.4.4': 'Deut.8.3',
    'Matt.21.42': 'Ps.118.22-Ps.118.23',
    'Rom.1.17': 'Hab.2.4'
}

mask = False
for from_verse, to_verse in pairs.items():
    mask |= (
        data['From Verse'].str.contains(from_verse) 
        & data['To Verse'].str.contains(to_verse)
    )

select_sample = pd.concat([data[mask], data.sample(7, random_state=42)])
select_sample = select_sample.sample(frac=1, random_state=42).reset_index(drop=True)
select_sample
```

```python
select_sample.to_csv('data/data_pilot_study/select_sample.tsv', sep='\t', index=False)
```

### Discussion Test 3a

Despite the same prompt, the outputs vary significantly. First, the model did not include `Rationale` as a column; this might be something worth explicitly including, as a safeguard against inaccurate classification. Second, out of the 10 edges, three differed. In all three cases, the edges were previously classified as `References to Specific Words or Phrases`. A few takeaways:
1. the model does not appear to fully distinguish between `References to Specific Words or Phrases` and `Comparative References`. Those two classes naturally have overlap, as similar themes will have similar phrases and vice versa.
2. It will be interesting to see if the model can accurately classify `Quoted References` (the class we are most interested in) with a high degree of recall.
